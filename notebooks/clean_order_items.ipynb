```python
import pandas as pd
import os
from ydata_profiling import ProfileReport

# Define relative path
report_dir = '../reports/documents'
report_filename = 'order_items_data_profile.html'
report_path = os.path.join(report_dir, report_filename)
```

```python
# Load the data
try:
    order_items_df = pd.read_csv('../data/raw/order_items.csv')
    print("File loaded successfully!")
except FileNotFoundError as e:
    print(e)
    print("make sure the files are in data/raw")
```

```python
# Use the profile report
if 'order_items_df' in locals():
    order_items_profile = ProfileReport(order_items_df, title="Order Items Data Profile")
    # Turn the profile report into an HTML file
    order_items_profile.to_file(report_path)
    print("Profile report is successfully created.")
```
