Question: What are the top 5 best-selling product category by revenue?
1. We need product category - Product Dataset
2. We need the total price for each category - Order Items Dataset
3. (Optional) Product category name in English - Category Name Dataset


```python
import pandas as pd
import os

# Define relative paths
data_dir = '../data/processed'
products_filename = 'products_clean.parquet'
order_items_filename = 'order_items_clean.parquet'
orders_filename = 'orders_clean.parquet'
products_path = os.path.join(data_dir, products_filename)
order_items_path = os.path.join(data_dir, order_items_filename)
orders_path = os.path.join(data_dir, orders_filename)
```

```python
# Load the parquet files
products_df = pd.read_parquet(products_path)
order_items_df = pd.read_parquet(order_items_path)
orders_df = pd.read_parquet(orders_path)
```

```python
# Merge the to table for analysis
merged_df = pd.merge(
    order_items_df,
    products_df,
    on='product_id',
    how='left'
)

# Now we group by category and sum by price to narrow down our analysis
category_revenue = merged_df.groupby('product_category_name')['price'].sum()

# Finally, we sort by price and return the top 5
top_5_categories = category_revenue.sort_values(ascending=False).head(5)

# Print the result
print(top_5_categories)
```

Question: "How have the sales of our top 5 product categories trended over time?"

```python
# Merge with the orders table to get the date information
final_df = merged_df.merge(
    orders_df,
    on='order_id',
    how='left'
)

final_df.head()

# Save the merged dataframe to a CSV file for Tableau
final_df.to_csv('../data/processed/tableau_product_revenue.csv', index=False)
```
