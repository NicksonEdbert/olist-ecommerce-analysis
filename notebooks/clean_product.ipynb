## Imports and Setup

```python
import pandas as pd
import os
from ydata_profiling import ProfileReport

# Define relative path
report_dir = '../reports/documents'
report_filename = 'product_data_profile.html'
report_path = os.path.join(report_dir, report_filename)
```

## Functions
### Function for Cleaning Product

```python
# Load the data
try:
    products_df = pd.read_csv('../data/raw/products.csv')

    # Rows without product category is an invalid records and can't be used for analysis, so we drop them.
    print(f'Number of rows before cleaning: {len(products_df)}')
    products_df_cleaned = products_df.dropna(subset='product_category_name')
    print(f'Remaining rows after cleaning: {len(products_df_cleaned)}')

    print("File loaded successfully!")
except FileNotFoundError as e:
    print(e)
    print("make sure the files are in data/raw")
```

```python
# Use the profile report
if 'products_df_cleaned' in locals():
    product_profile = ProfileReport(products_df_cleaned, title="Products Data Profile")
    # Turn the profile report into an HTML file
    product_profile.to_file(report_path)
    print("Profile report is successfully created.")

```
