```python
import pandas as pd
import os
from ydata_profiling import ProfileReport

# Define relative path
report_dir = '../reports/documents'
report_filename = 'orders_data_profile.html'
report_path = os.path.join(report_dir, report_filename)
```

```python
# Load the data
try:
    orders_df = pd.read_csv('../data/raw/orders.csv')

    # Fill the missing delivery dates with Nat(Not a time) instead of deleting the entire row to preserve data integrity
    orders_df['order_delivered_customer_date'] = pd.to_datetime(orders_df['order_delivered_customer_date'], errors='coerce')
    orders_df['order_delivered_carrier_date'] = pd.to_datetime(orders_df['order_delivered_carrier_date'], errors='coerce')

    orders_df.info()

    print("File loaded successfully!")
except FileNotFoundError as e:
    print(e)
    print("make sure the files are in data/raw")
```

```python
# Use the profile report
if 'orders_df' in locals():
    orders_profile = ProfileReport(orders_df, title="Order Data Profile")
    # Turn the profile report into an HTML file
    orders_profile.to_file(report_path)
    print("Profile report is successfully created.")
```
